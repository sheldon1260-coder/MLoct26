```python
selection = int(input("Please seslect an option (1-4): "))
match selection:
    case 1:
        print("You selected option 1")
    case 2:
        print("You selected option 2")
    case 3:
        print("You selected option 3")
    case 4:
        print("You selected option 4")
    case _:
        print("You selected an invalid option" )
```
